```python
import pickle
import os

esmc_dir = "embeddings/esmc_600m_features"
saprot_dir = "embeddings/saprot_650m_features"

# Extract UniProt IDs from filenames
esmc_ids = {f.split(" ")[0]: f for f in os.listdir(esmc_dir)}
saprot_ids = {f.replace(".pkl", ""): f for f in os.listdir(saprot_dir)}

# Find common IDs
common_ids = set(esmc_ids.keys()) & set(saprot_ids.keys())
print(f"Common IDs: {len(common_ids)}")

# Check shapes for all common IDs
for uid in sorted(common_ids):
    with open(f"{esmc_dir}/{esmc_ids[uid]}", "rb") as f:
        esmc = pickle.load(f)
    with open(f"{saprot_dir}/{saprot_ids[uid]}", "rb") as f:
        saprot = pickle.load(f)
    
    esmc_shape = esmc.shape if hasattr(esmc, 'shape') else len(esmc)
    saprot_shape = saprot.shape if hasattr(saprot, 'shape') else len(saprot)
    
    print(f"{uid}: ESMC={esmc_shape}, SaProt={saprot_shape}")
```
